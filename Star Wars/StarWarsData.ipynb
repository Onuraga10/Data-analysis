```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('StarWars.csv', encoding="ISO-8859-1")
```

```python
data
```

```python
datac=data.copy()
```

```python
datac.drop(index=0,inplace=True)
```

```python
datac
```

```python
datac.isna().sum()
```

```python
datac.columns
```

Cleaning and Mapping Yes/No Columns

```python
yesnoq=['Have you seen any of the 6 films in the Star Wars franchise?',
        'Do you consider yourself to be a fan of the Star Wars film franchise?',
        'Are you familiar with the Expanded Universe?',
        'Do you consider yourself to be a fan of the Expanded Universe?æ',
        'Do you consider yourself to be a fan of the Star Trek franchise?']

yes_no= {
    'Yes': True,
    'No':False
}
for i in yesnoq:
    datac[i] = datac[i].map(yes_no)
```

```python
datac[yesnoq]
```

Cleaning and Mapping Checkbox Columns

```python
names = ['seen_1','seen_2','seen_3','seen_4','seen_5','seen_6',]
a=0
mapper = {}

for i in datac.columns[3:9]:
    mapper.update({i:names[a]})
    a+=1
    
datac.rename(columns=mapper,inplace=True)
```

```python
for i in datac.columns[3:9]:
    datac[i] = datac[i].notna()

datac[datac.columns[3:9]]
```

 Cleaning the Ranking Columns

```python
datac[datac.columns[9:15]]= datac[datac.columns[9:15]].astype(float)

datac[datac.columns[9:15]].dtypes
```

```python
names2 = ['ranking_1','ranking_2','ranking_3','ranking_4','ranking_5','ranking_6',]
a=0
mapper2 = {}

for i in datac.columns[9:15]:
    mapper2.update({i:names2[a]})
    a+=1
datac.rename(columns=mapper2,inplace=True)
datac[datac.columns[9:15]]
```

Finding the Highest-Ranked Movie

```python
average_ranking= datac[datac.columns[9:15]].mean()

plt.bar(datac.columns[9:15],6-average_ranking)

plt.xlabel('Movies')
plt.ylabel('Average Ranking')
plt.title('Ranking For Star Wars Movies')
```

We can see here that the fifth movie which is 'Star Wars: Episode V The Empire Strikes Back' was liked the most and third movie which is 'Star Wars: Episode III Revenge of the Sith' was liked the least.

Finding the Most Viewed Movie

```python
seen = datac[datac.columns[3:9]].sum()
```

```python
plt.bar(datac.columns[3:9],seen)
plt.title('Movie Views')
plt.xlabel('Film Number')
plt.ylabel('View Count')
```

We can see here that fifth star wars movie was the most seen movie out off all. Fifth movie was also the most liked movie out off all 6. So we can say that there is a correlation between rankings and views.

Exploring the Data by Binary Segments

Now I will do the same analyses but now splitting the data by genders

```python
males = datac2[datac2['Gender']=='Male']
females = datac2[datac2['Gender']=='Female']
```

```python
mranking= males[males.columns[9:15]].mean()
franking= females[females.columns[9:15]].mean()
```

```python
x_indexes= np.array(datac2.columns[9:15])

barWidth=0.50

plt.subplot(1,2,1)
plt.bar(x_indexes,mranking,width=barWidth)
plt.xticks(rotation=90)
plt.title('Male Ranking')
plt.xlabel('Movie Nunber')
plt.ylabel('Ranking')

plt.subplot(1,2,2)
plt.bar(x_indexes,franking,color='purple',width=barWidth)
plt.xticks(rotation=90)
plt.title('Female Ranking')
plt.xlabel('Movie Nunber')
plt.ylabel('Ranking')
plt.tight_layout()
```

```python
mviews= males[males.columns[3:9]].sum()
fviews= females[females.columns[3:9]].sum()
```

```python
x_indexes= np.array(datac2.columns[3:9])

barWidth=0.50

plt.subplot(1,2,1)
plt.bar(x_indexes,mviews,width=barWidth)
plt.xticks(rotation=90)
plt.title('Male Views')
plt.xlabel('Movie Nunber')
plt.ylabel('Views')

plt.subplot(1,2,2)
plt.bar(x_indexes,fviews,color='purple',width=barWidth)
plt.xticks(rotation=90)
plt.title('Female Views')
plt.xlabel('Movie Nunber')
plt.ylabel('Views')
plt.tight_layout()
```
