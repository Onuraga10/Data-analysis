```python
import pandas as pd
import os
from matplotlib import pyplot as plt
```

```python

raw_data = pd.read_csv('games.csv', index_col='id')
```

```python
raw_data.columns
```

```python
raw_data.drop_duplicates(inplace=True)
raw_data.duplicated().sum()
```

```python
print('White rating mean: ',int(raw_data['white_rating'].mean()))
print('Black rating mean: ',int(raw_data['black_rating'].mean()))
```

```python
print('Correlation between white rating and turns',raw_data['white_rating'].corr(raw_data['turns']))
print('Correlation between white rating and turns',raw_data['black_rating'].corr(raw_data['turns']))
```

```python
raw_data.groupby('victory_status').count()['rated']
```
